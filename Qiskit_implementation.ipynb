#Python Basics: Vectors and Matrices

- A qubit state is represented as a 2×1 column vector. In Qiskit we often use NumPy to build and manipulate these.

```python
'''defines two column vectors, ket0 and ket1, corresponding to the qubit state vectors
∣0⟩ and ∣1⟩, and then prints their average.'''

import numpy as np

ket0 = np.array([[1],[0]])  # |0⟩
ket1 = np.array([[0],[1]])  # |1⟩
print(np.average(ket0),np.average(ket1))
print("##############################")

print(ket0 / 2 + ket1 / 2 )
```

#States, measurements, and operations

# state vector:
- Qiskit offers the Statevector class, which represents quantum states cleanly.
 - state vector is simply a vector represinting the amplitude of |0⟩ and |1⟩.
- example : |+⟩ = 1/√2 (|0⟩ + |1⟩)
     - is a valid qubit state because the squares of the  amplitudes add up to 1.

```python
from qiskit.quantum_info import Statevector
from numpy import sqrt
u= Statevector([1/sqrt(2), 1/sqrt(2)])
v= Statevector([(1+2j)/3, -2/3])
w= Statevector([1/3, 2/3])
```

```python
# foe display used display(u.draw(""))
display(u.draw("text"))
display(u.draw("latex"))
# display(u.draw("latex_source"))
print(f'##############')
display(v.draw("text"))
display(v.draw("latex"))
# display(v.draw("latex_source"))
```

```python

display(w.draw("text"))
print("\n")
display(w.draw("latex"))
print('\n')
display(w.draw("latex_source"))
```

```python
#check whether a state is valid by is_valid(),(sum of probabilities is 1)
# is_valid method which checks to see if a given vector is a valid quantum state vecto
# in other word that it has Euclidean norm equal to 1

display(u.is_valid())
display(v.is_valid())
display(w.is_valid())
```

```python
# 1
# Import the NumPy library for numerical operations
import numpy as np
# Import array_to_latex for displaying arrays in LaTeX format
from qiskit.visualization import array_to_latex
```

```python
#2
# Define ket0 (|0⟩) and ket1 (|1⟩) state vectors as NumPy arrays

ket0 = np.array([[1], [0]])  # Represents the quantum state |0⟩  or constant 0 vector
ket1 = np.array([[0], [1]])  # Represents the quantum state |1⟩  or constant 1's vector
```

```python
#3
# Define a superposition state (e.g., |+⟩ = 1/√2 (|0⟩ + |1⟩))
plus_ket = (ket0 + ket1) / np.sqrt(2) # Calculate the superposition state |+⟩

```

```python


print("Displaying state vectors:")
print("|0⟩:") # Label for the |0⟩ state
display(array_to_latex(ket0)) # Display the |0⟩ state vector using LaTeX rendering

print("|1⟩:") # Label for the |1⟩ state
display(array_to_latex(ket1)) # Display the |1⟩ state vector using LaTeX rendering

print("|+\u27E9 = ( |0\u27E9 + |1\u27E9 ) / \u221A2 :") # Label for the |+⟩ state
display(array_to_latex(plus_ket)) # Display the |+⟩ state vector using LaTeX rendering
```

```python

# Convert numpy arrays to Statevector objects for validation

state_ket0 = Statevector(ket0.flatten())
# flatten to 1D array for Statevector
state_ket1 = Statevector(ket1.flatten())
 # flatten to 1D array for Statevector
state_plus_ket = Statevector(plus_ket.flatten())
 # flatten to 1D array for Statevector

print("Is |0⟩ a valid state?", state_ket0.is_valid())
print("Is |1⟩ a valid state?", state_ket1.is_valid())
print("Is |+⟩ a valid state?", state_plus_ket.is_valid())
```

#operations with Operator and Statevector
- unary operations can be defined by operator class
- In quantum computing, all gates are unitary i.e. meaning they preserve probability

```python
from qiskit.quantum_info import Operator
Y = Operator([[0,-1.0j], [1.0j, 0]])
H = Operator([[1/sqrt(2),1/sqrt(2)], [1/sqrt(2),-1/sqrt(2)]])
S = Operator([[1, 0], [0, 1.0j]])
T = Operator([[1, 0], [0, (1 + 1.0j) / sqrt(2)]])
display(T .draw("latex"))
```

```python
#apply a unitary operation to a state vector using the evolve method
V = Statevector([1,0])

V = v.evolve(H)
V = v.evolve(Y)
V = v.evolve(S)
V = v.evolve(T)
display(V.draw("latex"))
```

```python

v = v.evolve(H)
v = v.evolve(T)
v = v.evolve(H)
v = v.evolve(S)
v = v.evolve(Y)

display(v.draw("latex"))
```

eg2 : with operator class

```python
from qiskit.quantum_info import Operator

# Pauli-Y matrix
Y = Operator([[0, -1.0j], [1.0j, 0]])

# Hadamard gate
H = Operator([[1/sqrt(2), 1/sqrt(2)],
              [1/sqrt(2), -1/sqrt(2)]])
```

evolve (apply) operators to states:

```python
state = Statevector([1,0]) # state in |0>
state = state.evolve (H) # applies Hadamard to |0⟩, creating a superposition
display(state.draw("latex"))
```

```python
from qiskit.quantum_info import Statevector, Operator
from numpy import sqrt
from qiskit import QuantumCircuit
from qiskit.visualization import plot_histogram

# Create a superposition state
psi = Statevector([1/sqrt(2), 1/sqrt(2)])
print("Initial State:", psi.draw("text"))

# Simulate measurement
counts = psi.sample_counts(500)
print("Counts:", counts)
plot_histogram(counts)

# Apply more gates via a circuit
qc = QuantumCircuit(1)
qc.h(0)
qc.s(0)
state_after = psi.evolve(qc)
print("After gates:", state_after.draw("text"))
qc.draw(output='mpl')
```

another example :in this case, will simply be a sequence of unitary operations performed on a single qubit

```python
from qiskit import QuantumCircuit

circuit = QuantumCircuit(1)

circuit.h(0)
circuit.t(0)
circuit.h(0)
circuit.s(0)
circuit.y(0)

display(circuit.draw(output="mpl"))
```

```python
# by using from_circuit() methode
display(Operator.from_circuit(circuit).draw("latex"))
```

```python
ket0 = Statevector([1, 0])  # initilize a starting quantum vector
v = ket0.evolve(circuit)
display(v.draw("latex"))
```

```python
statistics = v.sample_counts(4000)
display(plot_histogram(statistics))
```
