```python
# import all need libery:
from qiskit.quantum_info import Statevector, Operator
from qiskit.visualization import plot_histogram
from numpy import sqrt
import matplotlib.pyplot as plt
```

```python
# 1. create executive uncertainity state.
def create_executive_state(alpha=1/sqrt(2), beta=1/sqrt(2)):
  """
    Creates a normalized executive decision state.

    alpha → weight of Safe strategy
    beta  → weight of Risk strategy

    Default = balanced uncertainty.
    """
  state = Statevector([alpha, beta])

  if not state.is_valid():
    raise ValueError("State is not normalized. Ensure alpha^2 + beta^2 = 1")

  return state
```

```python
# 2. define cognetive bias operators
def positive_market_signal():
    """
    Positive market news increases risk appetite.
    Phase boost to |1⟩ (Risk).
    """
    return Operator([
        [1, 0],
        [0, 1j]
    ])


def negative_market_signal():
    """
    Negative news increases safety preference.
    Phase flip of |1⟩.
    """
    return Operator([
        [1, 0],
        [0, -1]
    ])


def executive_stress_rotation():
    """
    Simulates emotional amplification using rotation.
    Increases probability weight dynamically.
    """
    return Operator([
        [1/sqrt(2), 1/sqrt(2)],
        [1/sqrt(2), -1/sqrt(2)]
    ])
```

```python
# 3: SIMULATE DECISION COLLAPSE:
def simulate_decision(state, shots=1000):
    """
    Measures executive decision multiple times
    to observe strategic bias pattern.
    """
    counts = state.sample_counts(shots)

    print("\nDecision Interpretation:")
    print("0 → SAFE STRATEGY")
    print("1 → RISK STRATEGY")
    print("Measurement Counts:", counts)

    plot_histogram(counts)
    plt.title("Executive Strategic Bias Distribution")
    plt.show()
```

```python
# STEP 4: COMPLETE EXECUTION FLOW
if __name__ == "__main__":

    print("Initializing Executive Decision Engine...\n")

    # Balanced uncertainty
    executive_state = create_executive_state()

    print("Initial Strategic State:")
    print(executive_state.draw("text"))

    # Apply market condition
    biased_state = executive_state.evolve(positive_market_signal())

    # Optional emotional amplification
    biased_state = biased_state.evolve(executive_stress_rotation())

    print("\nState After Cognitive Influence:")
    print(biased_state.draw("text"))

    # Collapse decision
    simulate_decision(biased_state)
```

```python
# Quantum Cognitive Decision Engine (QCDE) - Interactive Version

# User dynamically inputs:
# - Strategic confidence
# - Market condition
# - Emotional amplification

# This version is fully interactive and will display the measurement histogram correctly.


from qiskit.quantum_info import Statevector, Operator
from qiskit.visualization import plot_histogram
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# STEP 1: User-defined uncertainty
# -----------------------------
def get_confidence_input():
    while True:
        try:
            safe_prob = float(input("Enter SAFE strategy confidence (0 to 1): "))
            if 0 <= safe_prob <= 1:
                break
            else:
                print("Value must be between 0 and 1.")
        except:
            print("Invalid input. Enter a decimal number.")

    risk_prob = 1 - safe_prob

    # Convert probabilities to amplitudes
    alpha = np.sqrt(safe_prob)
    beta = np.sqrt(risk_prob)

    return alpha, beta

# -----------------------------
# STEP 2: Market bias operator
# -----------------------------
def get_market_operator():
    print("\nSelect Market Condition:")
    print("1 → Positive Market (encourages risk)")
    print("2 → Negative Market (encourages safety)")
    print("3 → Neutral (no bias)")

    choice = input("Enter option (1/2/3): ")

    if choice == "1":
        return Operator([[1, 0], [0, 1j]])
    elif choice == "2":
        return Operator([[1, 0], [0, -1]])
    else:
        return Operator([[1, 0], [0, 1]])

# -----------------------------
# STEP 3: Emotional amplification
# -----------------------------
def get_emotional_rotation():
    print("\nEmotional Intensity Level:")
    print("1 → Low")
    print("2 → Medium")
    print("3 → High")

    level = input("Select intensity (1/2/3): ")

    if level == "1":
        theta = np.pi / 8
    elif level == "2":
        theta = np.pi / 4
    else:
        theta = np.pi / 2

    # Simple rotation matrix to simulate interference
    return Operator(
        [[np.cos(theta), np.sin(theta)], [np.sin(theta), -np.cos(theta)]]
    )

# -----------------------------
# STEP 4: Simulate measurement & show histogram
# -----------------------------
def simulate(state, shots=1000):
    # Sample measurement counts
    counts = state.sample_counts(shots)

    # Convert NumPy types to standard Python types for plotting
    clean_counts = {str(k): int(v) for k, v in counts.items()}

    print("\nDecision Outcome:")
    print("0 → SAFE STRATEGY")
    print("1 → RISK STRATEGY")
    print("Measurement Counts:", clean_counts)

    # Plot histogram
    fig = plot_histogram(clean_counts)
    plt.title("Executive Strategic Decision Distribution")
    plt.show()  # Explicitly show plot

# -----------------------------
# MAIN EXECUTION
# -----------------------------
if __name__ == "__main__":
    print("=== Quantum Executive Decision Simulator ===\n")

    # 1. User enters initial uncertainty
    alpha, beta = get_confidence_input()
    state = Statevector([alpha, beta])

    print("\nInitial Decision State (superposition):")
    print(state.draw("text"))

    # 2. Apply market bias
    market_operator = get_market_operator()
    state = state.evolve(market_operator)

    # 3. Apply emotional amplification
    emotion_operator = get_emotional_rotation()
    state = state.evolve(emotion_operator)

    print("\nFinal Cognitive State Before Measurement:")
    print(state.draw("text"))

    # 4. Simulate measurement and display histogram
    simulate(state)
```
